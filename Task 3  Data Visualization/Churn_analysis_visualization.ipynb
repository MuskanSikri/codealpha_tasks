```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('churn_data.csv')
```

```python
df
```

```python
df["Churn"].value_counts().plot(kind="pie", autopct='%1.1f%%')
plt.title("Customer Churn Distribution")
plt.ylabel("")
plt.show()
```

```python
#This pie chart represents the overall churn distribution. Approximately XX% of customers have churned, while XX% have retained the service. This indicates the current retention performance of the company
```

```python
sns.countplot(x="gender", hue="Churn", data=df)
plt.title("Churn by Gender")
plt.show()
```

```python
#This countplot shows churn distribution across gender. It helps us understand whether gender has any significant impact on customer churn
```

```python
sns.countplot(x="Contract", hue="Churn", data=df)
plt.xticks(rotation=45)
plt.title("Churn by Contract Type")
plt.show()
```

```python
#This visualization shows that customers with month-to-month contracts have a significantly higher churn rate compared to one-year and two-year contract customers. This suggests that long-term contracts help improve customer retention.
```

```python
sns.boxplot(x="Churn", y="MonthlyCharges", data=df)
plt.title("Monthly Charges vs Churn")
plt.show()
```

```python
#This boxplot compares monthly charges between churned and non-churned customers. We can observe that customers who churned generally had higher monthly charges. This indicates that pricing may influence customer retention
```

```python
sns.boxplot(x="Churn", y="tenure", data=df)
plt.title("Tenure vs Churn")
plt.show()
```

```python
#This boxplot shows that customers with lower tenure are more likely to churn. Long-term customers tend to stay with the company. This indicates that customer loyalty increases over time
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
#This heatmap shows the correlation between numeric variables. We observe that tenure and total charges have a strong positive correlation, which is expected as customers who stay longer accumulate higher total charges. Other variables show moderate or weak correlations.
```
