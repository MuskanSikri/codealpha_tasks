```python
def conv(value):
    if value ==1:
        return "yes"
    else:
        return "no"

df['SeniorCitizen']=df['SeniorCitizen'].apply(conv)
    
```

```python
ax= sns.countplot(x= 'Churn' , data=df)
ax.bar_label(ax.containers[0])
plt.title("Count of customers by churn")
plt.show()
```

```python
plt.figure(figsize= (3,4))
gb=df.groupby("Churn").agg({'Churn':"count"})
plt.pie(gb["Churn"], labels =gb.index, autopct= "%1.2f%%")
plt.title("Percentage of churned customers" , fontsize=10)
plt.show()
```

```python
# from the given pie chart we can conclude that 26.54 % of our customers have churned out
#now lets explore the reason behind it
```

```python
#the data is not gender specificso we can conclude that the data is not affected by gender
sns.countplot(x= "gender", data =df ,hue = 'Churn')
plt.title("Churn by gender")
plt.show()
```

```python
sns.countplot(x= "SeniorCitizen", data =df ,hue = 'Churn')
plt.title("Churn by SeniorCitizen")
plt.show()
```

```python

# Step 1: Groupby count
grouped = df.groupby(['SeniorCitizen', 'Churn']).size().unstack()

# Step 2: Convert to percentage within each SeniorCitizen group
grouped_percent = grouped.div(grouped.sum(axis=1), axis=0) * 100

# Step 3: Plot stacked bar
ax = grouped_percent.plot(kind='bar', stacked=True, figsize=(6,5), colormap='Set2')

plt.title("Churn by Senior Citizen Status (% within group)")
plt.xlabel("Senior Citizen (0 = No, 1 = Yes)")
plt.ylabel("Percentage")

# Step 4: Add percentage labels
for i in range(len(grouped_percent)):
    cumulative = 0
    for value in grouped_percent.iloc[i]:
        ax.text(i, cumulative + value/2,
                f"{value:.1f}%",
                ha='center', va='center', color='black', fontsize=9)
        cumulative += value

plt.legend(title="Churn")
plt.xticks(rotation=0)
plt.show()
```

```python


# List of categorical columns (from what you shared)
columns = [
    'gender', 'SeniorCitizen', 'Partner', 'Dependents',
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

# Set figure size
plt.figure(figsize=(30, 50))

# Create subplots
for i, col in enumerate(columns):
    plt.subplot(5, 3, i+1)   # 5 rows, 3 columns grid
    sns.countplot(x=col, data=df , hue=df['Churn'])
    plt.title(col , fontsize=30)
    plt.xticks(fontsize=40)

plt.tight_layout()
plt.show()
```

```python
#The count plots show that most customers are non-senior citizens, do not have dependents, and have phone and internet services, with Fiber Optic being common. Customers without add-on services like Online Security, Tech Support, and Device Protection appear more frequently in the churn category.

#Churn is higher among customers with month-to-month setups (implied by service patterns) and those lacking support services, while customers with additional security and support features tend to churn less.
```
